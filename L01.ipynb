## Chapter 1: Basics of Numerical Analysis {#sec-chapter-1}

- Absolute and relative error
- Rate of convergence
- Using plots to show rate of convergence

Suppose we approximate $x$ with $\widetilde{x}$. Here, $x$ could be e.g. a root (or fixed point) of a function, the value of an integral, the solution of an ODE or PDE evaluated at a point, .... .   

::: {#def-error}
# Errors

We define:

\begin{align}
\text{Actual error:} &\quad x - \widetilde{x} \\
\text{Absolute error:} &\quad |x - \widetilde{x}|
\end{align}

And when $x \neq 0$, the relative error is:

\begin{align}
\text{Relative error:} \quad \frac{|x - \widetilde{x}|}{|x}
\end{align}

:::

::: {#exm-errors}

If $x = 1.5$ and $\widetilde{x} = 1$, then the absolute error is $0.5$ and the relative error is $\frac{0.5}{1.5} = \frac13$.   

If $x = 1.5 \times 10^{42}$ and $\widetilde{x} = 10^{42}$, then the absolute error is $0.5 \times 10^{42}$ and the relative error is $\frac13$.  

If $x = 1.5 \times 10^{-42}$ and $\widetilde{x} = 10^{-42}$, then the absolute error is $0.5 \times 10^{-42}$ and the relative error is $\frac13$.

:::

The relative error is a better measure for small or large $x$:

::: {#exm-Planck}

Approximating Planck's constant $6.62607015 \times 10^{-34} \text{m}^2 \text{kg } \text{s}^{-1}$ with $0 \text{m}^2 \text{kg } \text{s}^{-1}$ would give an absolute error of $\approx 6 \times 10^{-34}$ (which seems small) but a relative error of $1$. In practice $0$ is a bad approximation to Planck's constant!

:::

::: {#def-acc-digits}
# Number of accurate digits

Suppose $x \not= 0$ is approximated by $\widetilde{x}$. Then, we say this approximation is accurate to 

$$\eta := -\log_{10} \left| \frac{x-\widetilde{x}}{x} \right| $$

digits.

:::

::: {#exm-e}

Approximating $e := \lim_{n\to\infty} \big( 1 + \frac1n \big)^n \approx 2.7182818.....$ with $3, 2.7,$ and $2.718$ gives $\eta = 0.98, 2.17$, and $3.98$ (to $2$ decimal places), respectively.

:::

We now consider a sequence of approximations $x_n$ converging to $x$. That is, for all $\varepsilon > 0$, there exists $N=N(\varepsilon)$ such that $|x - x_n| < \varepsilon$ for all $n > N$. In numerical analysis, we are interested in *(i)* whether our numerical schemes converge, and *(ii)* the rate of convergence: 

::: {#def-rate-convergence}
# Rate of convergence

Suppose $\varepsilon_n \geq 0$ and $(\varepsilon_n) \to 0$. Then, $x_n \to x$ with rate $O(\varepsilon_n)$ if there exist $C$ and $N$ such that

$$
|x - x_n| \leq C\varepsilon_n
$$

for all $n > N$. We also write $x_n = x + O(\varepsilon_n)$.

:::

::: {#exr-rate-1}

Show that $\frac{n + \cos(n)}{n^2} = O(\frac1n)$ and $\frac{n^2+3}{n^{42}} = O(\frac1{n^{40}})$.

:::


When plotting graphs, $O(n^{-k})$ decay (that is, **algebraic** decay) can be seen as straight lines on log-log plots ($-k$ is the slope in such graphs):

````julia
N = 1:100;
plot( N.^(-1) ;
    xlabel="n", xaxis=:log, yaxis=:log, 
    linestyle = :dot, lw=2, label = L"n^{-1}")
plot!( N.^(-2); 
    linestyle = :dot, lw=2, label = L"n^{-2}")
plot!( N.^(-5); 
    linestyle = :dot, lw=2, label = L"n^{-5}")
````

On the other hand, $O(e^{-c N})$ (that is, **exponential** decay) can be seen as straight lines on a semi-log y plot (and $-c$ is the gradient):

````julia
plot( exp.(-N);
    xlabel="n", yaxis=:log, 
    linestyle = :dot, lw=2, label = L"e^{-n}")
plot!( exp.(-2 .* N ); 
    linestyle = :dot, lw=2, label = L"e^{-2n}")
plot!( exp.(-5 .* N ); 
    linestyle = :dot, lw=2, label = L"e^{-5n}")
````
