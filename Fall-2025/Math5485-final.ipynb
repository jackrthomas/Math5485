6. **[5 points]** In the following, we plot the Lebesgue constants for interpolation in *(i)* Equispaced points on $[-1,1]$, and *(ii)* in Chebyshev nodes. Which set of constants grows faster as a function of $n$? Explain your answer with reference to $\textrm{(B.5)}$.

````julia
# | echo: false

function ChebyshevNodes( n; type=2) 
    if (type == 1)
        return @. cos( (2*(n:-1:0)+1)*pi/(2*(n+1)) )
    else
        return @. cos( (pi/n)*(n:-1:0) )
    end
end

function Lagrange( i, n; e="Chebyshev" ) 

    if (e=="Chebyshev")
        x = ChebyshevNodes(n)
    else 
        x = range(-1,1,n+1)
    end
    y = [ (j==i+1) ? 1. : 0. for j = 1:(n+1) ]

    if (e=="Chebyshev")
        λ = zeros(n+1)
        λ[1] = 1/2
        λ[n+1] = (-1)^n/2
        for j in 1:(n-1)
            λ[j+1] = (-1)^j
        end
    else 
        # equispaced...
        λ = [(-1)^j * binomial(n, j) for j ∈ 0:n]
    end

    return t -> t ∈ x ? ( (t==x[i+1]) ? 1. : 0. ) : sum( @. λ * y / (t - x) ) / sum( @. λ / (t - x) )
end

mesh = -1:.01:1

Λ, Λ′ = [], [] 
for n = 2:50
    f = x-> sum( [abs( Lagrange(i,n;e="equi")(x) ) for i in 0:n] )
    push!( Λ, maximum( f.(mesh) ) )

    g = x-> sum( [abs( Lagrange(i,n)(x) ) for i in 0:n] )
    push!( Λ′, maximum( g.(mesh) ) )
end

scatter( Λ, xlabel=L"n", yaxis=:log, title="Lebesgue constants", color="black", label="Equispaced points", legend=false )
scatter!( Λ′, label="Chebyshev nodes", color="red" )
# scatter!( ChebyshevNodes(n), [ (j==i+1) ? 1. : 0. for j = 1:(n+1) ] )
````

````julia
# | echo: false

nn = 2:5:150
mesh = range(-1+1e-10,1-1e-10,100)

a=10
F(x) = cos(a*x)# (1-x^2) * cos(a*x) 
F′(x) = -a * sin(a*x) #-2x * cos(a*x) + a*(-1 + x^2)*sin(a*x)

err = []
# push!( err, maximum( @. abs( f(mesh) - S(mesh) ) ) )

for n ∈ nn
    h = 2/n 
    X = -1:h:1
    Y = F.(X)

    A = diagm( -1 => fill(1,n-2), 0 => fill(4, n-1), 1 => fill(1, n-2) )
    b = [ 3*(Y[j+1]-Y[j-1])/h for j ∈ 2:n ]
    b[1] = b[1] - F′(-1)
    b[n-1] = b[n-1] - F′(1)
    m = A \ b

    e = 0.
    for x ∈ mesh
        S = 0;
        if (x <= -1+2/n)
            # x ∈ [x_0, x_1]
            t = (x-X[1])/h
            S = F(-1) * H0(t) + F(-1+2/n) * H1(t) + h * ( F′(-1.) * K0(t) + m[1] * K1(t) )
        elseif (x >= 1-2/n)
            t = (x-X[n])/h
            S = F(X[n]) * H0(t) + F(X[n+1]) * H1(t) + h * ( m[n-1] * K0(t) + F′(1.) * K1(t) )
        else
            idx = findlast(s->s<x, X)
            # X[idx] ≤ x ≤ X[idx+1] 
            t = (x - X[idx])/h
            S = F(X[idx]) * H0(t) + F(X[idx]+2/n) * H1(t) + h * ( m[idx-1] * K0(t) + m[idx] * K1(t) ) 
        end
        
        if ( abs( F(x) - S ) > e ) 
            e = abs( F(x) - S ) 
        end
    end

    push!( err, e )
end

scatter( nn, err, xaxis=:log, yaxis=:log, title="Error in cubic spline interpolation", legend=false, ticks=false, size=(500,300))
plot!( nn, 150 * (nn).^(-4), linestyle=:dash )
````

## Singular value decomposition (SVD)

Recall that the SVD of $A \in \mathbb R^{m\times n}$ is $A = P\Sigma Q^T$ where $P\in \mathbb R^{m\times m}$ is orthogonal ($PP^T = P^TP = I$), $\Sigma\in \mathbb R^{m\times n}$ is diagonal with *singular values* $\sigma_i := \Sigma_{ii}$, $\sigma_1 \geq \sigma_2 \geq \dots \geq \sigma_r \geq 0$, and $r := \min\{m,n\}$, and $Q \in \mathbb R^{n\times n}$ is orthogonal.   

5. **[4 points]** What are the singular values of the following matrix:

\begin{align}
    \begin{pmatrix}
        1 & 0 \\ 
        1 & 0 \\
        1 & 0
    \end{pmatrix}.\tag{D.3}
\end{align}

Recall that $\|A\| := \max_{\bm x\not= 0} \frac{|A\bm x|}{|\bm x|}$ where $|\bm x|:= \sqrt{\sum_{i} x_i^2}$. 

6. **[4 points]** Use the SVD of $A$, together with the fact that $|u|^2 = \sum_{i=1}^m u_i^2 = u^T u$, to show that 

\begin{align}

    = \max_{\bm y \not= 0} 
    \frac
        {|P\Sigma y|^2}
        {|y|^2} 
    %
    = \max_{\bm y \not= 0} \frac{\sum_{i} \sigma_i^2 y_i^2}{\sum_i y_i^2}.\tag{D.4}
\end{align}
