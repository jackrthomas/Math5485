Why do we use the residuals $r_n := |f(x_n)|$ instead of the errors $|x_n - \xi|$ when we are approximating the order of convergence ($\alpha$) in this method?

<div class='alert alert-block alert-success'><b>Answer.</b>

\begin{align}
    \lim_{n\to\infty} \frac{ \log |f(x_{n+1})| }{ \log |f(x_n)| } 
    %
    &= \lim_{n\to\infty}\frac{ \log |f'(\eta_{n+1})| +  \log |x_{n+1} - \xi| }{  \log |f'(\eta_{n})| +  \log |x_{n} - \xi| } \nonumber\\

    %
    &= \lim_{n\to\infty} \frac{\log |x_{n+1} - \xi|}{\log |x_{n} - \xi|}.
\end{align}

As a result, we may use the residuals to estimate the order of convergence.

## B: Midpoint Rule

Suppose $f$ is twice continuously differentiable. We will consider the following *midpoint* quadrature rule given by 

\begin{align*}
    \int_a^b f(x) \approx (b-a) f\big( \tfrac{a+b}2 \big).
\end{align*}

That is, there is only one node $x_0 = \tfrac{a+b}{2}$ with corresponding weight $w_0 = b-a$. 

1. **[3 points]** Show that this midpoint rule is exact for all polynomials of degree at most $1$,

<div class='alert alert-block alert-success'><b>Answer.</b> 

All polynomials of degree $1$ may be written in the form $f(x) = \alpha \big(x - \tfrac{a+b}{2}\big) + \beta$. By anti-symmetry about $\frac{a+b}2$, we have 

\begin{align}
    \int_a^b f &= \alpha \cdot 0 + \beta( b- a) \nonumber\\
    %
    &= (b - a) \Big( \alpha \big(\tfrac{a+b}2 - \tfrac{a+b}{2}\big) + \beta \Big) \nonumber\\
    %
    &= (b-a) f\big( \tfrac{a+b}{2} \big).
\end{align}

</div> 

2. **[3 points]** Write down the polynomial $p_1$ of degree at most $1$ with $p_1\big( \tfrac{a+b}{2} \big) = f\big( \tfrac{a+b}{2} \big)$ and $p_1'\big( \tfrac{a+b}{2} \big) = f'\big( \tfrac{a+b}{2} \big)$,

<div class='alert alert-block alert-success'><b>Answer.</b> 

\begin{align}
    p_1(x) =  f\big( \tfrac{a+b}{2} \big)  + f'\big( \tfrac{a+b}{2} \big) \big( x - \tfrac{a+b}2 \big) 
\end{align}

</div> 

3. **[3 points]** Show that there exists $\xi_x \in [a,b]$ such that  

\begin{align*}   
    f(x) - p_1(x) = \frac{f''(\xi_x)}{2} \big(x - \tfrac{a+b}2\big)^2.
\end{align*}

<div class='alert alert-block alert-success'><b>Answer.</b> 

By the Taylor remainder theorem, we have $f(x) = f(\tfrac{a+b}2) + f'(\frac{a+b}2) (x - \xi) + \frac12 f''(\xi_x) (x - \tfrac{a+b}2)^2 = p_1(x) + \frac12 f''(\xi_x) (x - \tfrac{a+b}2)^2$ for some $\xi_x$ between $x$ and $\tfrac{a+b}2$.

</div> 

4. **[8 points]** Show that

\begin{align*}
    \left|  \int_{a}^b f(x) \mathrm{d}x - (b-a) f\big( \tfrac{a+b}{2} \big) \right| \leq \frac{\| f'' \|_{L^\infty([a,b])}}{24} (b-a)^3 .
\end{align*}

<div class='alert alert-block alert-success'><b>Answer.</b> 

Since $p_1\big( \tfrac{a+b}{2} \big) = f\big( \tfrac{a+b}{2} \big)$ and $p_1$ is a polynomial of degree at most $1$ (and so the quadrature rule is exact for $p_1$), we have 

\begin{align}
    \left|  \int_{a}^b f(x) \mathrm{d}x - (b-a) f\big( \tfrac{a+b}{2} \big) \right|
    %
    &= \left|  \int_{a}^b f(x) \mathrm{d}x - (b-a) p_1\big( \tfrac{a+b}{2} \big) \right| \nonumber\\
    %
    &= \left|  \int_{a}^b f(x) \mathrm{d}x - \int_a^b p_1(x) \mathrm{d}x \right| \nonumber\\
    %
    &= \left|  \int_{a}^b \left( f(x) - p_1(x) \right) \mathrm{d}x \right| \nonumber\\
    %
    &= \left|\int_{a}^b  \frac{f''(\xi_x)}{2} \big(x - \tfrac{a+b}2\big)^2 \mathrm{d}x\right|   \nonumber\\
    %
    &\leq \frac{\|f''\|_{L^\infty([a,b])}}{2} \int_{a}^b \big(x - \tfrac{a+b}2\big)^2 \mathrm{d}x  \nonumber\\
    %
    &= \frac{\|f''\|_{L^\infty([a,b])}}{6} \big(x - \tfrac{a+b}2\big)^3\Big|_a^b  \nonumber\\
    %
    &= \frac{\|f''\|_{L^\infty([a,b])}}{6} \frac{2(b-a)^3}{8}.
\end{align}

</div> 


Let $z_{j} := a + j \frac{b-a}{n}$ for $j = 0,\dots,n$ and note that $\int_a^b f = \sum_{j=0}^{n-1} \int_{z_{j}}^{z_{j+1}} f$.

5. **[5 points]** Apply the midpoint rule on each interval $[z_j, z_{j+1}]$ to write down a formula for the corrresponding composite rule on $[a,b]$. Give the weights $\{w_j\}$ and nodes $\{x_j\}$ of this rule. 

<div class='alert alert-block alert-success'><b>Answer.</b> 

We find that 

\begin{align}
    \int_a^b f &= \sum_{j=0}^{n-1} \int_{z_{j}}^{z_{j+1}} f \nonumber\\
    %
    &\approx \sum_{j=0}^{n-1} \big( z_{j+1} - z_j \big) f\big( \tfrac{z_j + z_{j+1}}{2} \big) \nonumber\\
    %
    &= \sum_{j=0}^{n-1} \frac{b-a}{n} f\big( a + (2j+1) \tfrac{b-a}{2n} \big)
\end{align}

is a quadrature rule with nodes $x_j =  a + (2j+1) \frac{b-a}{2n}$ and weights $w_j = \frac{b-a}n$ for $j= 0,\dots,n-1$. 

</div>

6. **[4 points]** Use the error estimate from Question 3 obtain an error estimate for the composite rule.

<div class='alert alert-block alert-success'><b>Answer.</b> 

Applying the error estimate on $[z_j, z_{j+1}]$ gives an error of 

\begin{align}
    \frac{\| f'' \|_{L^\infty([z_j,z_{j+1}])}}{24} (z_{j+1}-z_j)^3 
    %
    &\leq \frac{\| f'' \|_{L^\infty([a,b])}}{24} \left( \tfrac{b-a}{n} \right)^3 .
\end{align}

Summing these errors for $j = 0,\dots,n-1$ gives 

\begin{align}
\left| \int_a^b f - \sum_{j=0}^{n-1} \frac{b-a}{n} f\big( a + (2j+1) \tfrac{b-a}{2n} \big) \right| 
%
&\leq \frac{\| f'' \|_{L^\infty([a,b])}}{24} \frac{(b-a)^3}{n^2} .
\end{align}

***Remark.*** The rate of convergence of the composite rectangular rule was $O(n^{-1})$ and so, by choosing the midpoint instead of the left- (or right-) hand end-points, we get faster convergence at no extra computational cost. 
</div>

4. **[3 points]** Show that $L := \max_{x \in I} |g'(x)| = \frac14$.

<div class='alert alert-block alert-success'><b>Answer.</b> 

We have $g'(x) = - \frac{1}{(1+x)^2}$ which is also decreasing. As a result, we have $|g'(x)| \leq \frac{1}{(1+1)^2} = \frac{1}{4}$ for all $x \in I$. This bound is attained by $x = 1$. 
</div> 

5. **[3 points]** Explain why this means the iteration $x_{n+1} = g(x_n)$ converges to $\xi$ for all $x_1 \in I$.

<div class='alert alert-block alert-success'><b>Answer.</b> 

We have shown that $g: I \to I$ is a contraction ($|g(x) - g(y)| = |g'(z)| |x - y| \leq \frac{1}4|x-y|$ for some $z \in I$) and so by the contraction mapping theorem, we have $x_{n+1} = g(x_n)$ converges to $\xi$ for all $x_1 \in I$.
</div> 

6. **[5 points]** What is the order of convergence? 
7. **[5 points]** What is the asymptotic error constant?

<div class='alert alert-block alert-success'><b>Answer.</b> 

There exists $\eta_n$ between $x_n$ and $\xi$ such that 

\begin{align}
    \frac{|x_{n+1} - \xi|}{|x_n - \xi|} = \frac{|g(x_{n}) - g(\xi)|}{|x_n - \xi|} = | g'(\eta_n) |.
\end{align}

As a result, taking the limit, we see that the order of convergence is $1$ with asymptotic error constant 

\begin{align}

\end{align}

4. **[8 points]** By expanding the numerator $g(x) := f'(x) (x - \xi) - 2 f(x)$ and denominator $h(x) := f'(x)$, show that there exist $\eta_n, \theta_n$ between $x_n$ and $\xi$ such that 

\begin{align*}
    x_{n+1} - \xi &= 
    \frac
        {\frac1{3!} g'''(\eta_n) (x_n - \xi)^3}
        {h'(\theta_n) (x_n - \xi) }
    %
    = \frac16
    \left[ \frac
        {f'''(\eta_n) +  f^{(4)}(\eta_n) (\eta_n - \xi) }
        {f''(\theta_n)}
    \right]
    (x_n - \xi)^2.
\end{align*} 

<div class='alert alert-block alert-success'><b>Answer.</b> 

We take derivatives of $g$:

\begin{align}
    g(x) &= f'(x) (x - \xi) - 2 f(x)\nonumber\\
    g'(x) &= f''(x) (x - \xi) - f'(x) \nonumber\\
    g''(x) &= f'''(x) (x - \xi) \nonumber\\
    g'''(x) &= f^{(4)}(x) (x - \xi) + f'''(x).
\end{align}

Notice that $g(\xi) = g'(\xi) = g''(\xi) = 0$ and so there exists $\eta_n$ between $x_n$ and $\xi$ such that

\begin{align}
    g(x_n) &= g(\xi) + g'(\xi) (x_n - \xi) + \frac{1}2 g''(\xi) (x_n - \xi)^2 + \frac{1}{6} g'''(\eta_n) (x_n - \xi)^3 \nonumber\\
    %
    &= \frac{1}{6} \Big[ f'''(\eta_n) +  f^{(4)}(\eta_n) (\eta_n - \xi) \Big] (x_n-\xi)^3.
\end{align}

Moreover, there exists $\theta_n$ such that $f'(x_n) = f'(\xi) + f''(\theta_n) (x_n - \xi) = f''(\theta_n) (x_n - \xi)$ and so 

\begin{align}
    x_{n+1} - \xi &= 
    \frac16
    \left[ \frac
        {f'''(\eta_n) +  f^{(4)}(\eta_n) (\eta_n - \xi) } 
        {f''(\theta_n)}
    \right]
    (x_n - \xi)^2
\end{align}
