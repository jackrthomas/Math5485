<script>
(function() {
  // Extract notebook file name from current page
  var path = window.location.pathname.replace(/\/$/, "");
  var page = path.split("/").pop().replace(/\.html$/, "") || "lecture01";
  
  // 1. Set Raw Notebook Download URL
  var rawNotebookUrl = "https://raw.githubusercontent.com/jackrthomas/Math5485/main/" + page + ".ipynb";
  var downloadBtn = document.querySelector(".btn-download-ipynb");
  if (downloadBtn) {
    downloadBtn.setAttribute("href", rawNotebookUrl);
  }

  // 2. Set Ask a Question URL with current page title & URL pre-filled
  var pageTitle = encodeURIComponent("[Question] " + (document.title || page));
  var currentUrl = encodeURIComponent(window.location.href);
  var issueBody = encodeURIComponent("**Reference Page:** " + decodeURIComponent(currentUrl) + "\n\n**Question/comment:**\n[Your question/query/suggestion/comment here]");
  
  var issueUrl = "https://github.com/jackrthomas/Math5485/issues/new?template=question.md&title=" + pageTitle + "&body=" + issueBody;
  
  var questionBtn = document.querySelector(".btn-ask-question");
  if (questionBtn) {
    questionBtn.setAttribute("href", issueUrl);
  }
})();
</script>
