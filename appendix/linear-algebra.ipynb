# Linear Algebra {.unnumbered}

This appendix collects some facts from linear algebra that we use in the lectures on in Chapter 5: Linear systems of equations.

Throughout, $A \in \mathbb R^{m\times n}$ is a real matrix with entries $A_{ij}$ (for $1\leq i\leq m$ and $1 \leq j \leq n$), $x \in \mathbb R^n$ are column vectors, $e_j$ is the $j^\text{th}$ standard basis vector (with $(e_j)_i = \delta_{ij}$), and $I$ is the identity matrix.

::: {.callout-note}
The proofs on this page are **non-examinable**: they are included for completeness (click on "Proof" to expand them).
:::

## Notation

Basics

- $A^\top$ is the *transpose* of $A$: $(A^\top)_{ij} = A_{ji}$. We have $(AB)^\top = B^\top A^\top$ (since $\big((AB)^\top\big)_{ij} = \sum_k A_{jk} B_{ki} = (B^\top A^\top)_{ij}$) and, if $A$ is invertible, $(A^{-1})^\top = (A^\top)^{-1}$ (since $(A^{-1})^\top A^\top = (AA^{-1})^\top = I$ and $A^\top (A^{-1})^\top = (A^{-1}A)^\top = I$),
- $x^\top y = \sum_{j} x_j y_j$ is the *dot product* (or *inner product*) of $x, y \in \mathbb R^n$. We say $x$ and $y$ are *orthogonal* if $x^\top y = 0$,
- $|x| := \sqrt{x^\top x} = \sqrt{\sum_j x_j^2}$ is the *Euclidean norm* of $x$. We say $x$ is a *unit vector* if $|x| = 1$, and $u_1, \dots, u_k$ are *orthonormal* if $u_i^\top u_j = 0$ for $i\not= j$ and $|u_j| = 1$ for all $j$,
- $\|A\| := \max_{x\not=0} \frac{|Ax|}{|x|}$ is the *operator norm* of $A$,

Sets: 

- $\ker A := \{ x \in \mathbb R^n : Ax = 0\}$ is the *kernel* (or *null space*) of $A$, 
- $\mathrm{range}\, A := \{Ax : x\in \mathbb R^n\}$ is its *range*: the span of the columns of $A$,\
- $\mathrm{null}\, A := \dim \mathrm{ker}\, A$ is the *nullity* of $A$, 
- $\mathrm{rank}\, A := \dim \mathrm{range}\, A$ is the *rank* of $A$: the number of linearly independent columns (by @thm-A-svd, this is also the number of linearly independent rows),

Spectrum:

- $(\lambda, v)$ is an *eigenpair* of a square matrix $A$ if $Av = \lambda v$ and $v\not=0$, 
- if the above is true, $\lambda$ is an *eigenvalue* and $v$ is an *eigenvector*,
- $\sigma(A)$ is the set of eigenvalues of $A$. This is the *spectrum* of $A$.

Matrices: 

- $A$ is *symmetric* if $A^\top = A$, 
- $A$ is *diagonal* if $A_{ij} = 0$ for $i\not=j$, 
- $A$ is *upper triangular* if $A_{ij} = 0$ for $i > j$, 
- $A$ is *lower triangular* if $A_{ij} = 0$ for $i < j$,
- $A$ is *unit lower triangular* if $A$ is *lower triangular* and $A_{ii} = 1$ for all $i$,
- A square matrix $Q$ is *orthogonal* if $Q^\top Q = I$ (that is, $Q^{-1} = Q^\top$).

## Matrix multiplication

Let
$$
A = \begin{pmatrix}
    a_1 & a_2 & \cdots & a_p
\end{pmatrix} \in \mathbb R^{m\times p}, 
\qquad 
B = \begin{pmatrix}
    b_1^\top \\ b_2^\top \\\vdots\\ b_p^\top
\end{pmatrix} \in \mathbb R^{p \times n}
$$
where $a_k$ and $b_k^\top$ is the $k^\text{th}$ column of $A$ and $k^\text{th}$ row of $B$, respectively. Then, for $x \in \mathbb R^p$, we have that 
$$
Ax = \sum_{k=1}^p x_k a_k
$$
is a linear combination of the columns of $A$. Moreover, the product of $A$ and $B$ is given by
$$
\begin{align}\nonumber
(AB)_{ij} = \sum_{k=1}^p A_{ik} B_{kj} 
%
= \sum_{k=1}^p (a_k)_i (b_k)_j
\end{align}
$$
That is, $AB = \sum_{k} a_k b_k^\top$.

::: {#exr-A-rank}

Show that, if $u \in \mathbb R^m$ and $v\in \mathbb R^n$ are non-zero, then $u v^\top$ has rank $1$.

::: {.exercise-solution style="display: none;"}

We have $(uv^\top)x = (v^\top x)\, u$, and so the range of $uv^\top$ is $\mathrm{span}\{u\}$, which is one dimensional.

:::

:::

## Invertible matrices

::: {#thm-A-rank-nullity name="Rank–nullity theorem"}

For $A \in \mathbb R^{m\times n}$, we have $\mathrm{rank}\, A + \dim \ker A = n$.

:::

::: {.proof collapse="true"}

Let $v_1, \dots, v_k$ be a basis of $\ker A$ and extend it to a basis $v_1, \dots, v_n$ of $\mathbb R^n$. Since $Av_i = 0$ for $i \leq k$, every $Ax = A \sum_i c_i v_i = \sum_{i > k} c_i A v_i$ is in the span of $\{ A v_{k+1}, \dots, A v_n \}$, and so these vectors span $\mathrm{range}\, A$. 

They are also linearly independent: if $\sum_{i>k} c_i A v_i = 0$, then $\sum_{i > k} c_i v_i \in \ker A$, so $\sum_{i>k} c_i v_i = \sum_{i \leq k} d_i v_i$ for some $d_i$, and $c_i = 0$ for all $i$ since $v_1, \dots, v_n$ are linearly independent. Therefore, $\mathrm{rank}\, A = n - k = n - \dim\ker A$.

:::

We define the determinant of $A \in \mathbb R^{n\times n}$ by the *Leibniz formula*

$$
\begin{align}\nonumber
\det A := \sum_{\sigma \in S_n} \mathrm{sgn}(\sigma) \prod_{i=1}^n A_{i \sigma(i)},
\end{align}
$$

where $S_n$ is the set of permutations of $\{1, \dots, n\}$ and $\mathrm{sgn}(\sigma) \in \{\pm 1\}$ is the *sign* of $\sigma$. Recall that every permutation is a product of transpositions (permutations that swap two indices), $\mathrm{sgn}(\tau) = -1$ for a transposition $\tau$, and $\mathrm{sgn}(\sigma \circ \sigma') = \mathrm{sgn}(\sigma)\,\mathrm{sgn}(\sigma')$.

::: {#thm-A-det name="Properties of the determinant"}

For square $A, B \in \mathbb R^{n\times n}$ and $\alpha\in\mathbb R$,

1. $\det(AB) = \det A \det B$, and so $\det(A^{-1}) = \frac1{\det A}$ if $A$ is invertible,
2. $\det A^\top = \det A$ and $\det (\alpha A) = \alpha^n \det A$,
3. If $A$ is (upper or lower) triangular, then $\det A = \prod_{i=1}^n A_{ii}$,
4. Swapping two rows (or columns) changes the sign of $\det A$; adding a multiple of one row (or column) to another row (or column) does not change $\det A$,
5. If $A$ has a row (or column) of zeros, then $\det A = 0$.

:::

::: {.proof collapse="true"}

We first note three facts coming from Leibniz formula:

(a) after multiplying a row of $A$ by a constant $\alpha$, the determinant is also multiplied by $\alpha$ as each term in the sum contains exactly one entry of row $i$.

(b) Swapping two rows changes the sign of the determinant. Let $\tau$ be the transposition swapping rows $i$ and $j$, so the new matrix is $A'_{k\ell} = A_{\tau(k)\ell}$. Substituting $k = \tau(m)$ and $\sigma' := \sigma \circ \tau$ (which runs over $S_n$ as $\sigma$ does, with $\mathrm{sgn}(\sigma') = -\mathrm{sgn}(\sigma)$),
$$
\begin{align}\nonumber
\det A' = \sum_{\sigma} \mathrm{sgn}(\sigma) \prod_{m} A_{m\, \sigma(\tau(m))} = -\sum_{\sigma'} \mathrm{sgn}(\sigma') \prod_m A_{m \sigma'(m)} = -\det A.
\end{align}
$$

(c) If two rows of $A$ are equal, then swapping them does not change $A$, and so $\det A = -\det A$ by (b). That is, $\det A = 0$.

We now prove the statements, starting with 2.

**2.** Since $\mathrm{sgn}(\sigma)\,\mathrm{sgn}(\sigma^{-1}) = \mathrm{sgn}(\mathrm{id}) = 1$, we have $\mathrm{sgn}(\sigma^{-1}) = \mathrm{sgn}(\sigma)$, and $\sigma \mapsto \sigma^{-1}$ is a bijection of $S_n$. Therefore, substituting $j = \sigma(i)$,
$$
\begin{align}\nonumber
\det A^\top = \sum_\sigma \mathrm{sgn}(\sigma) \prod_i A_{\sigma(i) i} = \sum_\sigma \mathrm{sgn}(\sigma^{-1}) \prod_j A_{j \sigma^{-1}(j)} = \det A.
\end{align}
$$
Each term in the Leibniz formula is a product of $n$ entries, and so $\det(\alpha A) = \alpha^n \det A$. By $\det A^\top = \det A$, every statement about rows below also holds for columns.

**3.** Suppose $A$ is upper triangular. A term $\prod_i A_{i\sigma(i)}$ is non-zero only if $\sigma(i) \geq i$ for all $i$. Then $\sigma(n) = n$, and thus $\sigma(n-1) = n-1$, etc. and we must have $\sigma = \mathrm{id}$. Therefore $\det A = \prod_i A_{ii}$. For lower triangular $A$, apply this to $A^\top$.

**4.** Swapping two rows is (b). Adding $c$ times row $j$ to row $i$ gives, by (a), the determinant $\det A + c \det A''$ where $A''$ is $A$ with row $i$ replaced by row $j$. Since $A''$ has two equal rows, $\det A'' = 0$ by (c).

**5.** Each term in the Leibniz formula contains an entry of the zero row.

**1.** Fix $B$ and let $D(A) := \det(AB)$. Since $(AB)_{ij} = \sum_k A_{ik} B_{kj} = a_i^\top B$ where $a_{i}^\top$ is row $i$ of $A$. Therefore, (a)-(c) applied to $AB$ show that $D$ is linear in each row of $A$, changes sign when two rows of $A$ are swapped, and vanishes when two rows of $A$ are equal. Writing $a_i^\top = \sum_k A_{ik} e_k^\top$ and expanding using linearity in each row,
$$
\begin{align}\nonumber
    D(A) = \sum_{k_1, \dots, k_n} A_{1k_1} \cdots A_{nk_n}\, D(E_{k_1 \cdots k_n}),
\end{align}
$$
where $E_{k_1 \cdots k_n}$ has rows $e_{k_1}^\top, \dots, e_{k_n}^\top$. If two of the $k_i$ are equal, then $D(E_{k_1 \cdots k_n}) = 0$. Otherwise, $k_i = \sigma(i)$ for some $\sigma \in S_n$, and $E_{k_1\cdots k_n}$ is obtained from $I$ by swapping rows once for each transposition in $\sigma$, so $D(E_{k_1\cdots k_n}) = \mathrm{sgn}(\sigma) D(I)$. Therefore, $D(A) = \det A \cdot D(I) = \det A \det B$. Finally, if $A$ is invertible, then $\det A \det A^{-1} = \det I = 1$ (by 3).

:::

::: {#thm-A-invertible}

For a square matrix $A \in \mathbb R^{n\times n}$, the following are equivalent:

1. $A$ is invertible (there exists $A^{-1}$ with $A^{-1}A = AA^{-1} = I$),
2. For every $b \in \mathbb R^n$, the linear equation $Ax = b$ has exactly one solution,
3. $\det A \not= 0$,
4. $\ker A = \{0\}$ (i.e. $Ax = 0$ implies $x = 0$),
5. $\mathrm{rank}\, A = n$,
6. $\mathrm{null}\, A = 0$, 
7. The columns of $A$ are linearly independent, 
8. We have $0 \not\in \sigma(A)$ (i.e. $0$ is not an eigenvalue of $A$).

:::

::: {.proof collapse="true"}

We write $a_1, \dots, a_n$ for the columns of $A$, so that $Ax = \sum_{k=1}^n x_k a_k$. First, we show that 4, 5, 6, 7 and 8 are all equivalent:

- (4 $\Leftrightarrow$ 6) By definition, $\mathrm{null}\, A = \dim \ker A$, and a subspace has dimension $0$ if and only if it is $\{0\}$.
- (4 $\Leftrightarrow$ 7) The columns are linearly *dependent* if and only if $\sum_k x_k a_k = 0$ for some $x \not= 0$, i.e. if and only if $Ax = 0$ for some $x \not= 0$.
- (4 $\Leftrightarrow$ 8) $0 \in \sigma(A)$ if and only if $Av = 0 \cdot v = 0$ for some $v \not= 0$.
- (5 $\Leftrightarrow$ 6) By @thm-A-rank-nullity, $\mathrm{rank}\, A = n - \mathrm{null}\, A$.

Next, we show $1 \Rightarrow 2 \Rightarrow 4 \Rightarrow 1$:

- (1 $\Rightarrow$ 2) $x = A^{-1}b$ is a solution since $A(A^{-1}b) = b$. If $Ax = Ay = b$, then $x - y = A^{-1}A(x - y) = A^{-1}(b - b) = 0$, so the solution is unique.
- (2 $\Rightarrow$ 4) Taking $b = 0$, the equation $Ax = 0$ has exactly one solution. Since $x = 0$ is a solution, it is the only one.
- (4 $\Rightarrow$ 1) By 5, $\mathrm{range}\, A$ is an $n$ dimensional subspace of $\mathbb R^n$, and so $\mathrm{range}\, A = \mathbb R^n$. Therefore, for each $j$, there exists $b_j$ with $A b_j = e_j$. The matrix $B := [b_1 | \cdots | b_n]$ then satisfies $AB = I$. Moreover, $A(BA - I) = (AB)A - A = 0$, so every column of $BA - I$ is in $\ker A = \{0\}$. That is, $BA = I$ and $B = A^{-1}$.

Finally, we show $1 \Rightarrow 3 \Rightarrow 4$:

- (1 $\Rightarrow$ 3) By @thm-A-det, $\det A \det A^{-1} = \det (AA^{-1}) = \det I = 1$, and so $\det A \not= 0$.
- (3 $\Rightarrow$ 4) Suppose that 4 is not satisfied: $Ax = 0$ for some $x \not= 0$. Choose $k$ with $x_k \not= 0$ and, for each $j \not= k$, add $\frac{x_j}{x_k}$ times column $j$ to column $k$. By @thm-A-det, this does not change the determinant. The new $k^\text{th}$ column is $a_k + \sum_{j\not=k} \frac{x_j}{x_k} a_j = \frac{1}{x_k} Ax = 0$, and so (again by @thm-A-det) $\det A = 0$. That is, 3 is not satisfied.

Therefore, all of the statements are equivalent.

:::

## Eigenvalues

The eigenvalues of $A \in \mathbb R^{n\times n}$ are the roots of the *characteristic polynomial* $p(\lambda) := \det(\lambda I - A)$, a monic polynomial of degree $n$ (since $Av = \lambda v$ for some $v \not=0$ if and only if $\ker(\lambda I - A) \not= \{0\}$, i.e. if and only if $\det(\lambda I - A) = 0$ by @thm-A-invertible). Therefore, $A$ has $n$ eigenvalues $\lambda_1, \dots, \lambda_n$ (counted with multiplicity), which may be complex, and

$$
\begin{align}
\det A &= \prod_{i=1}^n \lambda_i, \nonumber\\
\mathrm{Tr}\, A &:= \sum_{i=1}^n A_{ii} = \sum_{i=1}^n \lambda_i.
\end{align}
$$

::: {.proof collapse="true"}

By the Leibniz formula, $p(\lambda) = \sum_\sigma \mathrm{sgn}(\sigma) \prod_i \big(\lambda \delta_{i\sigma(i)} - A_{i\sigma(i)}\big)$. If $\sigma \not= \mathrm{id}$, then $\sigma(i) \not= i$ for at least two $i$, so the corresponding term is a polynomial of degree at most $n - 2$. Therefore,
$$
\begin{align}\nonumber
p(\lambda) = \prod_{i=1}^n (\lambda - A_{ii}) + (\text{degree} \leq n-2) = \lambda^n - (\mathrm{Tr}\, A)\, \lambda^{n-1} + \cdots,
\end{align}
$$
which is monic of degree $n$. By the fundamental theorem of algebra, $p(\lambda) = \prod_{i=1}^n (\lambda - \lambda_i)$, whose coefficient of $\lambda^{n-1}$ is $-\sum_i \lambda_i$, so $\mathrm{Tr}\, A = \sum_i \lambda_i$. Moreover, $(-1)^n \prod_i \lambda_i = p(0) = \det(-A) = (-1)^n \det A$ by @thm-A-det.

:::

::: {#thm-A-similar name="Similar matrices"}

If $S$ is invertible, then $A$ and $S^{-1}AS$ have the same eigenvalues. Moreover, $(\lambda, v)$ is an eigenpair of $A$ if and only if $(\lambda, S^{-1}v)$ is an eigenpair of $S^{-1}AS$.

:::

::: {.proof collapse="true"}

By @thm-A-det, $\det(\lambda I - S^{-1}AS) = \det\big(S^{-1}(\lambda I - A)S\big) = \det(S^{-1})\det(\lambda I - A) \det S = \det(\lambda I - A)$. Also, $S^{-1}AS (S^{-1}v) = S^{-1}Av$, which equals $\lambda S^{-1}v$ if and only if $Av = \lambda v$.

:::

## Triangular and permutation matrices

::: {#thm-A-triangular name="Triangular matrices"}

1. If $L$ and $M$ are lower triangular, then $LM$ is lower triangular, 
2. If $L$ and $M$ are unit lower triangular, then $LM$ is unit lower triangular, 
3. A triangular matrix $L$ is invertible if and only if its diagonal entries are all non-zero, 
4. If $L$ is invertible and lower triangular, then $L^{-1}$ is triangular (and unit lower triangular if $L$ is),
5. The eigenvalues of a lower triangular matrix are its diagonal entries.

The same holds with "lower" replaced by "upper".

:::

::: {.proof collapse="true"}

We have $(LM)_{ij} = \sum_k L_{ik} M_{kj}$ and a term in this sum is non-zero only if $k \leq i$ and $j \leq k$. Therefore, $(LM)_{ij} = 0$ if $j > i$, and $(LM)_{ii} = L_{ii} M_{ii}$. If $L$ and $M$ are unit triangular, then the diagonal is $(LM)_{ii} = 1 \cdot 1 = 1$ and $LM$ is unit triangular. 

By @thm-A-det, $\det L = \prod_i L_{ii}$, which is non-zero if and only if every $L_{ii} \not= 0$. By @thm-A-invertible, this is the case if and only if $L$ is invertible. In this case, the $j^\text{th}$ column $x$ of $L^{-1}$ solves $Lx = e_j$. Forward substitution gives $x_1 = \cdots = x_{j-1} = 0$ and $x_j = \frac1{L_{jj}}$, and so $L^{-1}$ is lower triangular with diagonal entries $\frac1{L_{jj}}$.

$\lambda I - L$ is triangular with diagonal entries $\lambda - L_{ii}$, and so $\det(\lambda I - L) = \prod_i (\lambda - L_{ii})$. The roots of this equation are exactly the eigenvalues of $L$.

:::

A *permutation matrix* $P$ is the identity matrix with its rows reordered: $P$ has exactly one $1$ in each row and each column and zeros elsewhere. If the $i^\text{th}$ row of $P$ is $e_{p(i)}^T$, then $(PA)_{ij} = A_{p(i) j}$: multiplying by $P$ on the left reorders the rows of $A$. Permutation matrices are orthogonal: the columns of $P$ are distinct standard basis vectors, so $(P^\top P)_{ij} = e_{p(i)}^\top e_{p(j)} = \delta_{ij}$ (where now $e_{p(j)}$ is the $j^\text{th}$ column), i.e. $P^\top P = I$ and $P^{-1} = P^\top$.

## Orthogonal matrices

::: {#thm-A-orthogonal name="Orthogonal matrices"}

1. $Q \in \mathbb R^{m\times n}$ has orthonormal columns if and only if $Q^\top Q = I$.
2. If $m = n$, and $Q^\top Q = I$, then $QQ^T = I$, 
3. If $Q$ has orthonormal columns, then $(Qx)^T(Qy) = x^Ty$ and $|Qx| = |x|$ for all $x, y$,
4. The product of orthogonal matrices is orthogonal, and if $Q$ is orthogonal then so is $Q^T = Q^{-1}$,
5. If $Q$ is orthogonal then $\det Q = \pm 1$ and every eigenvalue $\lambda$ of $Q$ has $|\lambda| = 1$.

:::

::: {.proof collapse="true"}

We have $(Q^\top Q)_{ij} = q_i^\top q_j$ where $q_j$ is the $j^\text{th}$ column of $Q$. If $m = n$ and $Q^\top Q = I$, then $Qx = 0$ implies $x = Q^\top Q x = 0$, so $Q$ is invertible by @thm-A-invertible, and $Q^{-1} = (Q^\top Q) Q^{-1} = Q^\top$. Therefore, $QQ^\top = QQ^{-1} = I$.

We have $(Qx)^T(Qy) = x^TQ^TQy = x^Ty$ and, taking $y = x$, gives $|Qx| = |x|$.

We have $(Q_1Q_2)^T(Q_1Q_2) = Q_2^T Q_1^T Q_1 Q_2 = Q_2^TQ_2 = I$, and $(Q^T)^TQ^T = QQ^T = I$.

Since $\det I = 1$, we have $1 = \det(Q^TQ) = (\det Q)^2$. If $Qv = \lambda v$ with $v \in \mathbb C^n$ non-zero, then $|\lambda|^2 \bar v^\top v = \overline{(Qv)}^\top Qv = \bar v^\top Q^\top Q v = \bar v^\top v$ (as $Q$ is real), so $|\lambda| = 1$.

:::

::: {#thm-A-gram-schmidt name="Orthonormal bases"}

Any orthonormal vectors $u_1, \dots, u_k \in \mathbb R^m$ can be extended to an orthonormal basis $u_1, \dots, u_m$ of $\mathbb R^m$. Equivalently, any matrix with orthonormal columns can be extended (by adding columns) to an orthogonal matrix.

:::

## Symmetric matrices

::: {#thm-A-spectral name="Spectral theorem"}

A real matrix $A \in \mathbb R^{n\times n}$ is symmetric if and only if there exist an orthogonal matrix $P$ and a (real) diagonal matrix $\Lambda$ such that

$$
\begin{align}\nonumber
A = P \Lambda P^T.
\end{align}
$$

The diagonal entries $\lambda_i := \Lambda_{ii}$ are the eigenvalues of $A$ and the columns $p_1, \dots, p_n$ of $P$ are an orthonormal basis of $\mathbb R^n$ consisting of eigenvectors of $A$. That is, $A = \sum_{i=1}^n \lambda_i\, p_i p_i^T$.

:::

In particular, if $A$ is symmetric then

- every eigenvalue of $A$ is real,
- eigenvectors corresponding to different eigenvalues are orthogonal: if $Au = \lambda u$ and $Av = \mu v$ with $\lambda \not= \mu$, then $\lambda u^Tv = (Au)^Tv = u^TAv = \mu u^Tv$ and so $u^Tv = 0$.

More generally, a (possibly complex) matrix $A$ is *normal* (i.e. $A^\star A = AA^\star$, where $A^\star := \bar A^T$ is the conjugate transpose) if and only if $A = U\Lambda U^\star$ for some *unitary* $U$ (i.e. $U^\star U = I$) and diagonal $\Lambda$ (which may now be complex).

::: {#def-A-positive-definite name="Positive definite matrices"}

A symmetric matrix $A \in \mathbb R^{n\times n}$ is *positive semi-definite* if $x^TAx \geq 0$ for all $x\in\mathbb R^n$, and *positive definite* if $x^TAx > 0$ for all $x\not=0$.

:::

By @thm-A-spectral, $x^TAx = (P^Tx)^T \Lambda (P^Tx) = \sum_i \lambda_i (p_i^Tx)^2$, and so a symmetric matrix is positive semi-definite (respectively definite) if and only if all of its eigenvalues are $\geq 0$ (respectively $> 0$).

::: {.proof collapse="true"}

If every $\lambda_i \geq 0$, then $x^TAx = \sum_i \lambda_i (p_i^Tx)^2 \geq 0$. If every $\lambda_i > 0$ and $x \not= 0$, then $P^Tx \not= 0$ (as $P^T$ is invertible), so some $p_i^Tx \not= 0$ and $x^TAx > 0$. Conversely, taking $x = p_i$ gives $x^TAx = \lambda_i$.

:::

::: {#exm-A-ATA}

For any $A\in\mathbb R^{m\times n}$, the matrix $A^TA \in \mathbb R^{n\times n}$ is symmetric (since $(A^TA)^T = A^T (A^T)^T = A^TA$) and positive semi-definite since $x^TA^TAx = |Ax|^2 \geq 0$. Therefore, its eigenvalues are real and $\geq 0$. Moreover, $\ker(A^TA) = \ker A$: if $A^TAx = 0$, then $|Ax|^2 = x^TA^TAx = 0$ and so $Ax = 0$ (and conversely $Ax = 0$ implies $A^TAx = 0$). Therefore, by @thm-A-invertible, $A^TA$ is invertible if and only if $\ker A = \{0\}$, i.e. if and only if $A$ has linearly independent columns (as $Ax = \sum_k x_k a_k$). In this case, $A^TA$ is positive definite since $x^TA^TAx = |Ax|^2 > 0$ for $x \not= 0$.

:::

## Norms

::: {#thm-A-cauchy-schwarz name="Cauchy–Schwarz inequality"}

For all $x, y \in \mathbb R^n$, we have $|x^Ty| \leq |x|\,|y|$, with equality if and only if $x$ and $y$ are parallel.

:::

::: {.proof collapse="true"}

If $y = 0$ there is nothing to prove. Otherwise, for all $t\in\mathbb R$,

$$
\begin{align}\nonumber
0 \leq |x - ty|^2 = |x|^2 - 2t\, x^Ty + t^2 |y|^2.
\end{align}
$$

Choosing $t = x^Ty/|y|^2$ gives $0 \leq |x|^2 - (x^Ty)^2/|y|^2$, with equality if and only if $x = ty$.

:::

As a consequence, the Euclidean norm is a *norm*: $|x| \geq 0$ with $|x| = 0$ if and only if $x = 0$, $|\alpha x| = |\alpha|\,|x|$ and (the triangle inequality)

$$
\begin{align}\nonumber
|x + y|^2 = |x|^2 + 2x^Ty + |y|^2 \leq \big(|x| + |y|\big)^2.
\end{align}
$$

::: {#thm-A-operator-norm name="Properties of the operator norm"}

For $A, B$ of compatible sizes,

1. $\|A\| = \max_{|x| = 1} |Ax|$ and $|Ax| \leq \|A\|\,|x|$ for all $x$,
2. $\|\cdot\|$ is a norm on $\mathbb R^{m\times n}$, $\|I\| = 1$ and $\|AB\| \leq \|A\|\,\|B\|$,
3. if $P$ and $Q$ are orthogonal, then $\|PAQ\| = \|A\|$,
4. $|\lambda| \leq \|A\|$ for every eigenvalue $\lambda$ of a square matrix $A$,
5. if $A$ is symmetric with eigenvalues $(\lambda_i)$, then $\|A\| = \max_i |\lambda_i|$,
6. $\|A^T\| = \|A\|$ and $\|A^TA\| = \|A\|^2$. Therefore, $\|A\| = \sqrt{\lambda_{\max}(A^TA)}$, where $\lambda_{\max}(A^TA) \geq 0$ is the largest eigenvalue of $A^TA$ (see @exm-A-ATA and 5).

:::

## Singular value decomposition

::: {#thm-A-svd name="Singular value decomposition"}

Every $A\in\mathbb R^{m\times n}$ can be written as $A = P\Sigma Q^T$ where $P\in\mathbb R^{m\times m}$ and $Q\in\mathbb R^{n\times n}$ are orthogonal and $\Sigma\in\mathbb R^{m\times n}$ is diagonal with entries $\sigma_1 \geq \sigma_2 \geq \cdots \geq \sigma_r \geq 0$ ($r := \min\{m,n\}$), the *singular values* of $A$. Moreover,

1. $\sigma_1^2, \dots, \sigma_r^2$ are the $r$ largest eigenvalues of $A^TA$ (and of $AA^T$),
2. $\mathrm{rank}\, A = \mathrm{rank}\, A^T$ is the number of non-zero singular values (so the number of linearly independent columns of $A$ is equal to the number of linearly independent rows),
3. $A = \sum_{j=1}^r \sigma_j u_j v_j^T$ where $u_j$ and $v_j$ are the $j^\text{th}$ columns of $P$ and $Q$,
4. $\|A\| = \sigma_1$ and, if $A$ is square and invertible, $\|A^{-1}\| = 1/\sigma_n$ and so the condition number $\kappa(A) := \|A\|\,\|A^{-1}\|$ (Lecture 14) is $\kappa(A) = \sigma_1/\sigma_n$.
