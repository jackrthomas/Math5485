2. **[5 points]** Note that $x := 1.001^{2026}$ and $y := e^{2026 \log 1.001}$ are mathematically identical ($x=y$).

```julia
b, n, x = Float16( 1.001 ), 2026, Float16( 1 );
for i ∈ 1:n
    x = b*x
end
y = exp( n * log( b ) ) 
println( "x = ", x, "; y = ", y )
```

Which result do you trust more and why?
